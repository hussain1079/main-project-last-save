```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["revenue","revenue_source","employees_source","net_income_source","market_cap_source","enterprise_group_entity_status"])
y = df["revenue"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)


```

```python
cap_cols = [
    "employees",
    "employees_year",
    "revenue_year",
    "net_income",
    "net_income_year",
    "market_cap",
    "legal_units_count",
    "direct_subsidiaries_count",
    "max_hierarchy_depth",
]

# Keep only columns that exist in X_train
cap_cols = [col for col in cap_cols if col in X_train.columns]

lower_limits = X_train[cap_cols].quantile(0.01)
upper_limits = X_train[cap_cols].quantile(0.99)

X_train[cap_cols] = X_train[cap_cols].clip(
    lower=lower_limits,
    upper=upper_limits,
    axis=1,
)

# Apply the training limits to test data—don't calculate new test limits
X_test[cap_cols] = X_test[cap_cols].clip(
    lower=lower_limits,
    upper=upper_limits,
    axis=1,
)
```

```python
categorical_cols = [
    "enterprise_group_jurisdiction",
    "company_type",
    "revenue_currency",
    "net_income_currency",
    "market_cap_currency",
]


X_train_ready = X_train.drop(columns=["market_cap_date"])
X_test_ready = X_test.drop(columns=["market_cap_date"])

X_train_encoded = pd.get_dummies(
    X_train_ready,
    columns=categorical_cols,
    dtype=int
)

X_test_encoded = pd.get_dummies(
    X_test_ready,
    columns=categorical_cols,
    dtype=int
)

X_test_encoded = X_test_encoded.reindex(
    columns=X_train_encoded.columns,
    fill_value=0
)

print("Training encoded shape:", X_train_encoded.shape)
print("Testing encoded shape:", X_test_encoded.shape)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Original numeric features (do not include one-hot encoded columns)
numeric_features = [
    "employees",
    "employees_year",
    "revenue_year",
    "net_income",
    "net_income_year",
    "market_cap",
    "legal_units_count",
    "direct_subsidiaries_count",
    "max_hierarchy_depth",
    "market_cap_date_missing",
    "market_cap_year",
]

# Keep only columns that are actually present
numeric_features = [
    col for col in numeric_features
    if col in X_train_encoded.columns
]

# Impute using training data only
imputer = SimpleImputer(strategy="median")
X_train_scaled = X_train_encoded.copy()
X_test_scaled = X_test_encoded.copy()

X_train_scaled[numeric_features] = imputer.fit_transform(
    X_train_encoded[numeric_features]
)
X_test_scaled[numeric_features] = imputer.transform(
    X_test_encoded[numeric_features]
)

# Scale using training data only
scaler = StandardScaler()
X_train_scaled[numeric_features] = scaler.fit_transform(
    X_train_scaled[numeric_features]
)
X_test_scaled[numeric_features] = scaler.transform(
    X_test_scaled[numeric_features]
)

print("Training scaled shape:", X_train_scaled.shape)
print("Testing scaled shape :", X_test_scaled.shape)
print("Training numerical mean:",
      X_train_scaled[numeric_features].mean().mean())
print("Training numerical std:",
      X_train_scaled[numeric_features].std().mean())
```

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()

lr_model.fit(X_train_scaled, y_train)

y_pred = lr_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_encoded, y_train)
```

```python
import numpy as np
from sklearn.compose import TransformedTargetRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    median_absolute_error,
    mean_squared_error,
    r2_score,
)

model_log = TransformedTargetRegressor(
    regressor=LinearRegression(),
    func=np.log1p,
    inverse_func=np.expm1,
)

model_log.fit(X_train_scaled, y_train)
y_pred_log = model_log.predict(X_test_scaled)

print("MAE:", mean_absolute_error(y_test, y_pred_log))
print("Median absolute error:", median_absolute_error(y_test, y_pred_log))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred_log)))
print("R²:", r2_score(y_test, y_pred_log))
```
