```python
def detect_outliers(group):
    Q1 = group['Yield'].quantile(0.25)
    Q3 = group['Yield'].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    group = group.copy()

    group['is_outlier'] = (
        (group['Yield'] < lower_limit) |
        (group['Yield'] > upper_limit)
    )

    return group

df = df.groupby('Crop', group_keys=False).apply(detect_outliers)
```

```python
Q1 = df.groupby('Crop')['Yield'].transform('quantile', 0.25)
Q3 = df.groupby('Crop')['Yield'].transform('quantile', 0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df['is_outlier'] = (
    (df['Yield'] < lower_limit) |
    (df['Yield'] > upper_limit)
)
```

```python
from sklearn.model_selection import GroupShuffleSplit

X = df.drop(columns=['Yield', 'is_outlier', 'Year'])
y = df['Yield']


groups = (
    df['Region'].astype(str) + '_' +
    df['Year'].astype(str)
)

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()
```

```python
cap_cols = [
    'Nitrogen',
    'Phosphorus',
    'Potassium',
    'Fertilizer'
]


lower_limits = X_train[cap_cols].quantile(0.01)
upper_limits = X_train[cap_cols].quantile(0.99)


X_train.loc[:, cap_cols] = X_train[cap_cols].clip(
    lower=lower_limits,
    upper=upper_limits,
    axis=1
)

X_test.loc[:, cap_cols] = X_test[cap_cols].clip(
    lower=lower_limits,
    upper=upper_limits,
    axis=1
)

print("Capping completed!")
```

```python
numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=['object', 'category']
).columns.tolist()

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
for col in categorical_features:
    print(col, ":", X_train[col].nunique(), "unique categories")
```

```python

region_frequency = X_train['Region'].value_counts(normalize=True)


X_train['Region'] = X_train['Region'].map(region_frequency).fillna(0)
X_test['Region'] = X_test['Region'].map(region_frequency).fillna(0)
```

```python
numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = ['Crop', 'Season', 'Soil Type']

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
import pandas as pd

X_train_encoded = pd.get_dummies(
    X_train,
    columns=['Crop', 'Season', 'Soil Type'],
    dtype=int
)

X_test_encoded = pd.get_dummies(
    X_test,
    columns=['Crop', 'Season', 'Soil Type'],
    dtype=int
)


X_test_encoded = X_test_encoded.reindex(
    columns=X_train_encoded.columns,
    fill_value=0
)

print("Training encoded shape:", X_train_encoded.shape)
print("Testing encoded shape:", X_test_encoded.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()


X_train_scaled = X_train_encoded.copy()
X_test_scaled = X_test_encoded.copy()

X_train_scaled[numeric_features] = scaler.fit_transform(
    X_train_encoded[numeric_features]
)

X_test_scaled[numeric_features] = scaler.transform(
    X_test_encoded[numeric_features]
)

print("Training scaled shape:", X_train_scaled.shape)
print("Testing scaled shape :", X_test_scaled.shape)
print("Training numerical mean:",
      X_train_scaled[numeric_features].mean().mean())
print("Training numerical std:",
      X_train_scaled[numeric_features].std().mean())
```

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()

lr_model.fit(X_train_scaled, y_train)

y_pred = lr_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2 Score:", r2_score(y_test, y_pred))
```
