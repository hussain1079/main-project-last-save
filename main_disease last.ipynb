```python

selected_features = [
    'Crop',
    'Nitrogen',
    'Phosphorus',
    'Potassium',
    'pH_Value',
    'Soil_Temperature',
    'Humidity',
    'Rainfall',
    'Weather_Temp_F',
    'Weather_Humidity',
    'Cloud_Cover_pct',
    'Wind_Speed_kmh',
    'Rolling_Rainfall_mm'
]

# Separate input features and target
X = df[selected_features]
y = df['Disease']

print("Input features shape:", X.shape)
print("Target variable shape:", y.shape)
```

```python
print("Number of input features:", X.shape[1])
print("Number of disease classes:", y.nunique())

print("\nSelected features:")
print(X.columns.tolist())
```

```python


categorical_cols = X.select_dtypes(include=['object']).columns
numerical_cols = X.select_dtypes(include=np.number).columns

print("Categorical columns:", list(categorical_cols))
print("Numerical columns:", list(numerical_cols))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

print("\nTraining target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
print("Disease classes in training data:", y_train.nunique())
print("Disease classes in testing data:", y_test.nunique())

print("\nTraining disease distribution:")
display(y_train.value_counts().sort_index())

print("\nTesting disease distribution:")
display(y_test.value_counts().sort_index())
```

```python
numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=['object', 'category']
).columns.tolist()

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
X_train_encoded = pd.get_dummies(
    X_train,
    columns=categorical_features,
    dtype=int
)

X_test_encoded = pd.get_dummies(
    X_test,
    columns=categorical_features,
    dtype=int
)


X_test_encoded = X_test_encoded.reindex(
    columns=X_train_encoded.columns,
    fill_value=0
)

print("Training encoded shape:", X_train_encoded.shape)
print("Testing encoded shape:", X_test_encoded.shape)

display(X_train_encoded.head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = X_train_encoded.copy()
X_test_scaled = X_test_encoded.copy()

X_train_scaled[numeric_features] = scaler.fit_transform(
    X_train_encoded[numeric_features]
)

X_test_scaled[numeric_features] = scaler.transform(
    X_test_encoded[numeric_features]
)

print("Training scaled shape:", X_train_scaled.shape)
print("Testing scaled shape:", X_test_scaled.shape)
print(
    "Training numerical mean:",
    X_train_scaled[numeric_features].mean().mean()
)
print(
    "Training numerical standard deviation:",
    X_train_scaled[numeric_features].std().mean()
)
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

from sklearn.preprocessing import LabelEncoder


label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

model_results = {}
```

```python

from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42,
    n_jobs=-1
)


lr_model.fit(X_train_scaled, y_train)


y_pred_lr = lr_model.predict(X_test_scaled)


y_prob_lr = lr_model.predict_proba(X_test_scaled)


lr_accuracy = accuracy_score(y_test, y_pred_lr)

lr_precision = precision_score(
    y_test, y_pred_lr,
    average='macro',
    zero_division=0
)

lr_recall = recall_score(
    y_test, y_pred_lr,
    average='macro',
    zero_division=0
)

lr_f1 = f1_score(
    y_test, y_pred_lr,
    average='macro',
    zero_division=0
)

lr_roc_auc = roc_auc_score(
    y_test_encoded,
    y_prob_lr,
    multi_class='ovr',
    average='macro'
)


model_results['Logistic Regression'] = {
    'Accuracy': lr_accuracy,
    'Precision': lr_precision,
    'Recall': lr_recall,
    'F1 Score': lr_f1,
    'ROC-AUC': lr_roc_auc
}


print("Logistic Regression Evaluation")
print("--------------------------------")
print("Accuracy:", round(lr_accuracy, 4))
print("Precision:", round(lr_precision, 4))
print("Recall:", round(lr_recall, 4))
print("F1 Score:", round(lr_f1, 4))
print("ROC-AUC:", round(lr_roc_auc, 4))
```

```python
from sklearn.ensemble import RandomForestClassifier


rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_encoded, y_train)


y_pred_rf = rf_model.predict(X_test_encoded)


y_prob_rf = rf_model.predict_proba(X_test_encoded)


rf_accuracy = accuracy_score(y_test, y_pred_rf)

rf_precision = precision_score(
    y_test, y_pred_rf,
    average='macro',
    zero_division=0
)

rf_recall = recall_score(
    y_test, y_pred_rf,
    average='macro',
    zero_division=0
)

rf_f1 = f1_score(
    y_test, y_pred_rf,
    average='macro',
    zero_division=0
)

rf_roc_auc = roc_auc_score(
    y_test_encoded,
    y_prob_rf,
    multi_class='ovr',
    average='macro'
)


model_results['Random Forest'] = {
    'Accuracy': rf_accuracy,
    'Precision': rf_precision,
    'Recall': rf_recall,
    'F1 Score': rf_f1,
    'ROC-AUC': rf_roc_auc
}


print("Random Forest Evaluation")
print("--------------------------------")
print("Accuracy:", round(rf_accuracy, 4))
print("Precision:", round(rf_precision, 4))
print("Recall:", round(rf_recall, 4))
print("F1 Score:", round(rf_f1, 4))
print("ROC-AUC:", round(rf_roc_auc, 4))
```

```python
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from sklearn.ensemble import RandomForestClassifier
from scipy.stats import randint

# Create a stratified sample for hyperparameter tuning
X_tune, _, y_tune, _ = train_test_split(
    X_train_encoded,
    y_train,
    train_size=50000,
    stratify=y_train,
    random_state=42
)

print("Tuning dataset shape:", X_tune.shape)
print("Target shape:", y_tune.shape)
```

```python
param_grid = {
    'n_estimators': [100, 150, 200],
    'max_depth': [None, 20, 30, 40],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2']
}
```

```python
# Initialize Random Forest
rf = RandomForestClassifier(
    random_state=42,
    n_jobs=1
)

# Initialize Randomized Search
rf_random = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_grid,
    n_iter=5,
    cv=3,
    scoring='f1_macro',
    random_state=42,
    n_jobs=-1,
    verbose=2
)
```

```python
rf_random.fit(X_tune, y_tune)
```

```python
print("Best Parameters:")
print(rf_random.best_params_)

print("\nBest Cross-Validation F1 Score:")
print(rf_random.best_score_)
```
